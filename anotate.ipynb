```python
import os
from ultralytics import YOLO

# Load YOLOv8 model
model = YOLO('yolov8n.pt')

# Paths
images_dir = "train/images"
labels_dir = "train/labels"

# Create labels directory
os.makedirs(labels_dir, exist_ok=True)

# Supported image formats
image_extensions = [".jpg", ".jpeg", ".png"]

# COCO vehicle class IDs
vehicle_classes = [2, 3, 5, 7]

count_images = 0
count_labels = 0

# Walk through all folders (like rglob)
for root, dirs, files in os.walk(images_dir):
    for file in files:
        file_path = os.path.join(root, file)

        # Check extension
        if os.path.splitext(file)[1].lower() in image_extensions:
            print(f"Processing: {file}")
            count_images += 1

            # Run detection
            results = model(file_path, conf=0.25, classes=vehicle_classes)

            # Create label file path
            label_filename = os.path.splitext(file)[0] + ".txt"
            label_file = os.path.join(labels_dir, label_filename)

            lines = []

            for result in results:
                if result.boxes is None:
                    continue

                for box in result.boxes:
                    cls = 0  # vehicle class
                    x, y, w, h = box.xywhn[0].tolist()
                    lines.append(f"{cls} {x:.6f} {y:.6f} {w:.6f} {h:.6f}")

            # Write labels (even if empty)
            with open(label_file, "w") as f:
                f.write("\n".join(lines))

            if lines:
                count_labels += 1

print("\n Done labeling")
print(f"Images processed: {count_images}")
print(f"Images with detections: {count_labels}")
```
