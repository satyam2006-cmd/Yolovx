To function to check results from a list of students 

```python
score = [50,60,80,70,49]
def check_result(score):
  for i in score:
    if i <= 50:
       print(f"student {score.index(i)} has score {i} and Fail")
    else :
       print(f"student {score.index(i)} has score {i} and Pass")
check_result(score)
```
