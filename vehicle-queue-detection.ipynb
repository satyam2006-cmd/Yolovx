# Implementing Queue detection logic and Tunnig for Queue detection

```python
model = YOLO("/kaggle/working/runs/detect/train/weights/best.pt")
```

```python
# Video path
video_path = "/kaggle/input/datasets/satyambhagat/sample-video-for-testing-the-detection-of-vehicles/156157-811878168.mp4"
cap = cv2.VideoCapture(video_path)
```

```python
fps = int(cap.get(cv2.CAP_PROP_FPS))
if fps == 0:
    fps = 25
print("FPS:", fps)

# PARAMETERS
QUEUE_VEHICLE_THRESHOLD = 6
DISTANCE_THRESHOLD = 120
MIN_QUEUE_TIME = 2
MIN_QUEUE_FRAMES = int(fps * MIN_QUEUE_TIME)

ROI = (200, 100, 1000, 600)

queue_frame_count = 0

# DENSITY CHECK 
import numpy as np

def is_dense(boxes):
    if len(boxes) < 3:
        return False

    centers = [((x1+x2)/2, (y1+y2)/2) for x1,y1,x2,y2 in boxes]

    close_pairs = 0
    total_pairs = 0

    for i in range(len(centers)):
        for j in range(i+1, len(centers)):
            total_pairs += 1
            dist = np.linalg.norm(np.array(centers[i]) - np.array(centers[j]))
            if dist < DISTANCE_THRESHOLD:
                close_pairs += 1

    ratio = close_pairs / total_pairs if total_pairs > 0 else 0
    return ratio > 0.4   # tune this (0.3–0.6)

# SAVE OUTPUT
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter('output1.mp4', fourcc, fps, (1280, 720))

# MAIN LOOP 
while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    frame = cv2.resize(frame, (1280, 720))

    x1, y1, x2, y2 = ROI
    roi_frame = frame[y1:y2, x1:x2]

    # Using detection DETECTION (NOT TRACKING)
    results = model.predict(roi_frame)[0]

    boxes_np = []
    vehicle_count = 0

    if results.boxes is not None:
        boxes = results.boxes.xyxy.cpu().numpy()
        classes = results.boxes.cls.cpu().numpy()

        for box, cls in zip(boxes, classes):
            xA, yA, xB, yB = box
            cls = int(cls)

            # since you trained best.pt for vehicles → no need filtering
            boxes_np.append([xA, yA, xB, yB])
            vehicle_count += 1

            cv2.rectangle(roi_frame, (int(xA), int(yA)), (int(xB), int(yB)), (0,255,0), 2)

    # ---------------- QUEUE LOGIC ----------------
    density = is_dense(boxes_np)

    queue_condition = (
        vehicle_count >= QUEUE_VEHICLE_THRESHOLD and
        density
    )

    if queue_condition:
        queue_frame_count += 1
    else:
        queue_frame_count = max(0, queue_frame_count - 1)

    queue_detected = queue_frame_count >= MIN_QUEUE_FRAMES

    # DISPLAY
    status = "QUEUE DETECTED" if queue_detected else "Normal Traffic"

    cv2.putText(frame, f"Vehicles: {vehicle_count}", (20,40),
                cv2.FONT_HERSHEY_SIMPLEX, 1, (255,255,255), 2)

    cv2.putText(frame, f"Density: {density}", (20,80),
                cv2.FONT_HERSHEY_SIMPLEX, 1, (255,255,255), 2)

    cv2.putText(frame, f"Frames: {queue_frame_count}", (20,120),
                cv2.FONT_HERSHEY_SIMPLEX, 1, (255,255,255), 2)

    cv2.putText(frame, status, (20,160),
                cv2.FONT_HERSHEY_SIMPLEX, 1,
                (0,0,255) if queue_detected else (0,255,0), 3)

    cv2.rectangle(frame, (x1,y1), (x2,y2), (255,0,0), 2)

    frame[y1:y2, x1:x2] = roi_frame
    out.write(frame)

cap.release()
out.release()

print("Saved: output1.mp4")
```

```python
import cv2
import numpy as np
from ultralytics import YOLO
from collections import defaultdict
import math
from IPython.display import Video

# ---------------- LOAD ----------------
model = YOLO("/kaggle/working/runs/detect/train/weights/best.pt")
cap = cv2.VideoCapture("/kaggle/input/datasets/satyambhagat/traffic-analysis-video/car2lane.mp4")

# ---------------- VIDEO INFO ----------------
fps = int(cap.get(cv2.CAP_PROP_FPS))
if fps == 0:
    fps = 25

width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

print("FPS:", fps)

# ---------------- OUTPUT ----------------
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter("output2.mp4", fourcc, fps, (width, height))

# ---------------- PARAMETERS ----------------
STOP_SPEED = 2.0
STOP_FRAMES = int(fps * 2)
DIST_THRESHOLD = 50
MIN_QUEUE_VEHICLES = 4

# ---------------- STORAGE ----------------
next_id = 0
objects = {}  # id -> center
track_history = defaultdict(list)
stop_counter = defaultdict(int)

# ---------------- HELPER ----------------
def get_center(box):
    x1, y1, x2, y2 = box
    return (int((x1+x2)/2), int((y1+y2)/2))

def distance(p1, p2):
    return math.hypot(p1[0]-p2[0], p1[1]-p2[1])

# ---------------- MAIN LOOP ----------------
frame_count = 0

while True:
    ret, frame = cap.read()
    if not ret:
        print("Video Finished")
        break

    frame_count += 1
    detections = []

    # -------- DETECTION --------
    results = model(frame, conf=0.4)[0]

    if results.boxes is not None:
        boxes = results.boxes.xyxy.cpu().numpy()
        for box in boxes:
            detections.append(get_center(box))
    else:
        boxes = []

    # -------- TRACKING --------
    new_objects = {}

    for center in detections:
        matched = False

        for obj_id, prev_center in objects.items():
            if distance(center, prev_center) < DIST_THRESHOLD:
                new_objects[obj_id] = center
                matched = True
                break

        if not matched:
            new_objects[next_id] = center
            next_id += 1

    objects = new_objects

    # -------- SPEED + STOP DETECTION --------
    stopped_points = []

    for obj_id, center in objects.items():
        track = track_history[obj_id]
        track.append(center)

        if len(track) > 10:
            track.pop(0)

        if len(track) >= 2:
            dx = track[-1][0] - track[-2][0]
            dy = track[-1][1] - track[-2][1]
            speed = math.hypot(dx, dy)

            if speed < STOP_SPEED:
                stop_counter[obj_id] += 1
            else:
                stop_counter[obj_id] = 0

            if stop_counter[obj_id] > STOP_FRAMES:
                stopped_points.append(center)

    # -------- QUEUE DETECTION --------
    queue_detected = False

    if len(stopped_points) >= MIN_QUEUE_VEHICLES:
        xs = [p[0] for p in stopped_points]
        if np.std(xs) < 80:
            queue_detected = True

    # -------- DRAW --------
    for box in boxes:
        x1, y1, x2, y2 = map(int, box)
        cx, cy = get_center(box)

        color = (0, 255, 0)

        for p in stopped_points:
            if distance((cx, cy), p) < 20:
                color = (0, 0, 255)

        cv2.rectangle(frame, (x1, y1), (x2, y2), color, 2)

    if queue_detected:
        cv2.putText(frame, "QUEUE DETECTED", (30, 60),
                    cv2.FONT_HERSHEY_SIMPLEX, 1.5, (0, 0, 255), 3)
    else:
        cv2.putText(frame, "Normal Traffic", (30, 60),
                    cv2.FONT_HERSHEY_SIMPLEX, 1.5, (0, 255, 0), 3)

    # DEBUG INFO
    cv2.putText(frame, f"Stopped: {len(stopped_points)}", (30, 110),
                cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 2)

    cv2.putText(frame, f"Frame: {frame_count}", (30, 150),
                cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 0), 2)

    # -------- OUTPUT --------
    out.write(frame)

# ---------------- CLEANUP ----------------
cap.release()
out.release()

print("Done! Output saved as output2.mp4")
```
