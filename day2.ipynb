# Introduction to opencv and basic image processing

```python
import cv2

img = cv2.imread('image2.jpg')
if img is None:
 print("Image not found")
exit()

bright_img = cv2.convertScaleAbs(img, alpha=1, beta=80)

crop_img = img[50:200, 100:300]

pixels = img[100, 150]
print("Pixel value (BGR):", pixels)

cv2.imshow('Original Image', img)
cv2.imshow('Brightened Image', bright_img)
cv2.imshow('Cropped Image', crop_img)

cv2.imwrite('brightened_image.jpg', bright_img)
cv2.imwrite('cropped_image.jpg', crop_img)

cv2.destroyAllWindows()
cv2.waitKey(0)
```
