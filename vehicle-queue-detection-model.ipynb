```python
from ultralytics import YOLO
import torch, os

os.environ["PYTORCH_CUDA_ALLOC_CONF"] = "expandable_segments:True"
torch.cuda.empty_cache()

model = YOLO('yolov9c.pt')

results = model.train(
    data='/kaggle/working/Vehicle-queue-detection-1/data.yaml',
    epochs=50,
    imgsz=512,
    batch=8,
    device=[0,1],
    amp=True,
    workers=2,
    name='vehicle_queue_yolov9'
)
```
